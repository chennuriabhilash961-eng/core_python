Decorators

```python
# A Decorator is a function that takes an another function as input and adds some extra behaviour to it and returns the new enhanced function without changing the original functions code
# A Decorator is a wrapper function that modifies or extends another functions behaviour
```

```python
def greet():
    print("Hi")
say_hi=greet
say_hi()
```

```python
def outer():
    def inner():
        print("Hello")
    inner()
outer()
```

```python
def grett():
    print("hi")
    x=20
    return x
say_hi=grett()
print(say_hi)

print()
def greet():
    print("hi")
    x=20
    print(x)
    return x
say_hi=greet
value=say_hi()
print(value)
```

```python
def outer():
    def inner():
        print("Hi")
    inner()
    return inner
func=outer()
func()
```

```python
def outer():
    def inner():
        print("Hi")
    inner()
    return inner()
func=outer
func()
```

Function Decorators


```python
# Before using the fancy @syntax,lets build a deccorator manually step by step.This way you will truly understand what is happening under the hood.
```

```python
# Step 1 Functions are objects
# In python functions are first class objects you can store them in variables and pass them to other functions
def greet():
    print("Hello!")

say_hello=greet
say_hello()
```

```python
# Step 2 A Function inside a function (closure)
# A function can be defined in another function the inner function can access variables from the outer function's scope this is called a closure.
def outer():
    message='I am the outer function'
    def inner():
        print(message)
    inner()
outer()
```

```python
# step 3 Returning a function
# A function can return another function as its result .This is the heart of how decorators work
def make_greeter():
    def say_hi():
        print("Hi there!")
    return say_hi  # return the function itself
my_func=make_greeter()
my_func()
```
