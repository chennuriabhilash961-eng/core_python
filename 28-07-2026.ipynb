```python
#1 Given a list of integers,write a program to filter even numbers and then multiply each of them by 2 using a single pipeline of filter() and write your own version of map() called my_map(func,lst) using a regular loop
l=[2,3,7,8,22,56]
l1=list(filter(lambda x:x%2==0,l))
print(l1)
l2=list(map(lambda x:x*2,l1))
print(l2)

# Using single pipeline
l=[2,3,7,8,22,56]
print(list(map(lambda x:x*2,list(filter(lambda x:x%2==0,l)))))

def my_map(func,lst):
    s=[]
    for i in lst:
        x=func(i)
        s.append(x)
    return s
def square(x):
    return x*2
l=[2,3,5,7]
print(my_map(square,l))
```

```python
#2 write a function greet(name,prefix='Hello',formatter=lambda x:x)that returns a greeting message.Apply the formatter function to modify the output.Demonstrate the function using str.upper
def greet(name,prefix='Hello',formatter=lambda x:x):
    x=f"{prefix} {name}"
    return formatter(x)
print(greet("Abhi",prefix='Hi',formatter=lambda x:x.upper()))
print(greet("Abhi",formatter=lambda x:x.upper()))

```

```python
#3 Given a list of tuples(name,marks),write a program to sort the list by marks in descending order
l=[('charan',65),('ntr',75),('prabhas',98),('sujith',40)]
print(sorted(l,key=lambda x:x[1],reverse=True))
```

```python
#4 Given a list of product prices ,write a program to filter prices greater than 500,apply a 10% discount using map(),and find the total price using reduce()
from functools import reduce
l=[200,600,320,850,450,980]
l1=list(filter(lambda x:x>500,l))
print(l1)
l2=list(map(lambda x:x-(x*0.1),l1))
print(l2)
l3=reduce(lambda x,y:x+y,l2)
print(l3)

# Using Single pipeline solution
print(reduce(lambda x,y:x+y,list(map(lambda x:x-(x*0.1),list(filter(lambda x:x>500,l))))))
```
