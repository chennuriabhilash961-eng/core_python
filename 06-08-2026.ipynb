```python
# Step 4 Building a decorator
# Now combine everything: a function that takes a function ,define a wrapper inside, and returns the wrapper.
def my_decorator(func):
    def wrapper():
        print("Before the function returns")
        func()
        print("After the function returns")
    return wrapper
def say_hello():
    print("Hello World!")
say_hello=my_decorator(say_hello)
say_hello()
```

```python
# 1.     Create a function place_order(item)
#     Write a decorator that prints:
#     * “Function started” before execution
#     * “Function ended” after execution
def place_order(item):
    print("Item:",item)
    print("Function Started")
def decorator(func):
    def wrapper():
        func("pizza")
        print("Function ended")
    return wrapper
place=decorator(place_order)
place()
```
