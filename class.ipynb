```python
# Shorthand format
def decorator(fun):
    def wrapper(*args,**kwargs):
        print("before calling")
        fun(*args,**kwargs)
        print("After calling")
    return wrapper
@decorator
def add(a,b):
    print(a+b)
add(20,30)
```

```python
def shout(func):
    def wrapper(*args,**kwargs):
        s=func(*args,**kwargs)
        return str(s).upper()
    return wrapper
@shout
def greet(name):
    return f"hello, {name}"
print(greet("alice"))
```

```python
# 1.     Create a function get_message() that returns "hello user". Write a decorator using @ syntax that converts the output to uppercase.

def decorator(func):
    def wrapper(*args,**kwargs):
        x=func(*args,**kwargs)
        return str(x).upper()
    return wrapper
@decorator
def get_message():
    return f"hello, user"
print(get_message())
```

```python
# 2.     Create a function get_number() that returns 10
#     Use a decorator to return double the value.
def decorator(func):
    def wrapper(*args,**kwargs):
        x=func(*args,**kwargs)
        x=x*2
        return x
    return wrapper
@decorator
def get_number():
    return 10
print(get_number())
```
