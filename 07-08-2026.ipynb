### Decorators with *args and **kwargs


```python
# The wrapper above only words for function with no arguments To make a decorator work for any function always use *args and **kwargs in the wrapper
```

```python
def my_decorator(func):
    def wrapper(*args,**kwargs):
        print("before")
        func(*args,**kwargs)
        print("after")
    return wrapper
def add (a,b):
    print(a+b)
add=my_decorator(add)
add(4,5)
```

```python
def my_decorator(func):
    def wrapper(*args,**kwargs):
        print("before")
        l=func(*args,**kwargs)
        print("after")
        return l
    return wrapper
def add (a,b):
    return a+b
add=my_decorator(add)
print(add(4,5))
```

```python
def add(a,b):
    print(a+b)
def decorator(func):
    def wrapper():
        func(12,28)
    return wrapper
add=decorator(add)
add()
```

```python
def add(a,b):
    print(a+b)
def decorator(func):
    def wrapper():
        print("before addition")
        func(10,26)
        print("after addition")
    return wrapper
add=decorator(add)
add()
```

```python
def add(a,b):
    print(a+b)
def decorator(func):
    def wrapper(*args,**kwargs):
        print("before calling")
        func(*args,**kwargs)
        print("after calling")
    return wrapper
add=decorator(add)
add(23,75)
add(26,b=57)
add(26,b=92)
```

```python
def add(*args):
    sum=0
    for i in args:
        sum+=i
    print(sum)
def decorator(func):
    def wrapper(*args,**kwargs):
        print("before calling")
        func(*args,**kwargs)
        print("after calling")
    return wrapper
add=decorator(add)
add(12,3,4,55,6,6)
```
