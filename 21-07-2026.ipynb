```python
# Sorted
# sorted functions returns a new sorted list the key function determines what value to sort 
# syntax:sorted(iterable,key=function)
```

```python
students=[{'name':'Alice','score':78},
          {'name':'abhi','score':98},
          {'name':'Hari krishna','score':56}]
l=sorted(students,key=lambda x:x['score'])
print(l)
l1=sorted(students,key=lambda x:x['score'],reverse=True)
print(l1)
# based on length of names
print("length of names:",sorted(students,key=lambda x:len(x['name'])))
```

```python
# Q5. Chain map() and filter(): from [1 .. 10], first filter out odds, then square the remaining evens.
num=range(1,11)
l=list(filter(lambda x:x%2==0,num))
print(l)
l1=list(map(lambda x:x**2,l))
print(l1)
print(list(map(lambda x:x**2,list(filter(lambda x:x%2==0,num)))))
```

```python
# Q7. Use reduce() to find the longest string in a list: ['cat', 'elephant', 'dog', 'rhinoceros']
l=['cat','elephant','dog','rhinoceros']
l1=reduce(lambda x,y:x if len(x)>len(y) else y,l)
print(l1)
```

```python
# 1 Given a list of product prices, write a program to filter prices above ₹500, then apply a 10% discount using map(), and compute the final total bill using reduce().
from functools import reduce
l=[550,350,700,200,980]
l1=list(filter(lambda x:x>500,l))
print(l1)
l2=list(map(lambda x:x-(x*0.1),l1))
print(l2)
print(reduce(lambda x,y:x+y,l2))



```

```python
# 4 Given a list of words, write a program to filter words with length greater than 3, then convert them to uppercase using map(), and concatenate them into a single string using reduce().
words=['abhi','avinash','anu','arun']
l=list(filter(lambda x:len(x)>3,words))
print(l)
l1=list(map(lambda x:x.upper(),l))
print(l1)
l2=reduce(lambda x,y:x+' '+y,l1)
print(l2)
```

```python
# 8 ⁠A banking system stores transaction amounts. Write a program to filter only credit transactions (positive values), apply a processing bonus of ₹10 to each using map(), and calculate the total credited amount using reduce().
l=[-4000,2500,5000,-1000]
l1=list(filter(lambda x:x>0,l))
print(l)
l2=list(map(lambda x:x+10,l1))
print(l2)
l3=reduce(lambda x,y:x+y,l2)
print(l3)
# single pipe line solution
print(reduce(lambda x,y:x+y,list(map(lambda x:x+10,list(filter(lambda x:x>0,l))))))
```
