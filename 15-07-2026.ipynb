```python
# write a python program that creates a nested list and performs both shallow copy and deep copy operations
# Modify an inner element of the original list and display all lists Based on the output explain the
# difference between shallow copy and deep copy
a=[[23,34,45],[45,67,89],[98,65]]
print("original a:",a)
import copy
sc=copy.copy(a)
sc[0][2]=54
print("after shallowing the copy:",sc)
print("original of a",a)

# deepcopy
print("original of a",a)
dc=copy.deepcopy(a)
dc[2][0]=76
print("after deeping the copy:",dc)
print("original of a",a)
```

```python

# define a function calculate_total(price,tax=5,discount=0)that calculates the final price after adding tax
# and subtracting discount Demonstrate the function with different function calls to show her default
# arguments work
def calculate_total(price,tax=5,discount=0):
    total=price+tax
    total=total-discount
    return total
print(calculate_total(200))
print(calculate_total(500,20,10))
print(calculate_total(discount=30,price=300,tax=10))
```

```python
# write a function create_profile(name,*skills,**details) that accepts a persons name,multiple skills,and additional details
#such as age,city and experience Display the profile in a structured format
def create_profile(name,*skills,**details):
    print("Hi",name)
    print("your skills are:")
    for skill in skills:
        print(skill)
    print("Additional details:")
    for key,value in details.items():
        print(key,":",value)
create_profile("abhilash",
               "sql","java","python",
               age=23,city="kphb",experience="1 year")
```

```python
# create functions for addition ,subtraction and multiplication.store these functions in a dictionary
#and write a program that takes user input to select an operation and performs it dynamically using
# functional reference

def add(c,d):
    return c+d
def sub(c,d):
    return c-d
def mul(c,d):
    return c*d
method = {
    'addition':add,
    'subtraction':sub,
    'multiplication':mul
}
print(method["addition"](5,7))
operation=input("enter operation:")
print(method[operation](3,5))
```
