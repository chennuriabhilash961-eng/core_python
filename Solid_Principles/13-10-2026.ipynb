# SOLID Principles

```python
# The SOLID principles are five essential guidelines that enhance software design, making code more maintainable and scalable.
# 1.Single Responsibility Principle
# 2.Open Closed Principle
# 3.Liskov Substitution Principle
# 4.Interface Segregation Principle
# 5.Dependency Inversion Principle
```

# 1.Single Responsibility Principle


```python
# A class should have only one reason and should do exactly one thing

# This principle states that "A class should have only one reason to change" which means every class should have a single responsibility or single job or single purpose. In other words, a class should have only one job or purpose within the software system.


class User:
    def __init__(self,name,email):
        self.name=name
        self.email=email
class UserRepository:
    def save(self,user):    #only responsible for persistence
        pass
class EmailService:
    def send_welcome(self,user):        #only responsible for email
        pass
```

# 2.Open/Closed Principle


```python
# Open to extension
# closed to modification
# It defines that software entities should be open for extension but close for modification 
# In simple words we should be able to add new functionality with repeatedly changing existing code

# This principle states that "Software entities (classes, modules, functions, etc.) should be open for extension, but closed for modification" which means you should be able to extend a class behavior, without modifying it.

from abc import ABC,abstractmethod
class Discount(ABC):
    @abstractmethod
    def calculate(self,price):
        pass
class NoDiscount(Discount):
    def calculate(self,price):
        return price
class TenDiscount(Discount):
    def calculate(self,price):
        return price*0.90
class HalfPrice(Discount):      # New behaviour - no old code changed
    def calculate(self, price):
        return price*0.50
def checkout(price,discount:Discount):
    return discount.calculate(price)        #works with any discount sub class
print(checkout(1000,TenDiscount()))
print(checkout(1000,NoDiscount()))
print(checkout(1000,HalfPrice()))
```

# 3.Liskov Substitution Principle


```python
# 3.Liskov Substitution Principle
# A child class should be able to replace its parent without breaking the progrem expected output

# The principle was introduced by Barbara Liskov in 1987. According to this principle, "derived or child classes must be able to replace their base or parent classes". This ensures that any subclass can be used in place of its parent class without causing unexpected behavior in the program

from abc import ABC,abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        return self.length*self.breadth
class Square(Shape):
    def __init__(self,s):
        self.s=s
    def area(self):
        return self.s**2
obj=Square(5) 
obj.area()       
```

# 4. Interface Segregation Principle

```python
# 4. Interface Segregation Principle
# No class should be forced to implement methods it does not need
# Split large interfaces into smaller ones which are  specific a class should only implement what it actually needs.


from abc import ABC,abstractmethod
class Printable(ABC):
    @abstractmethod
    def print_doc(self):
        pass
class scannable(ABC):
    @abstractmethod
    def scan(self):
        pass
class BasicPrinter(Printable):
    def print_doc(self):
        return "Printing"
class Allinoneprinter(Printable,scannable):
    def print_doc(self):
        return "printing"
    def scan(self):
        return "SCanning"
    
```

# 5. Dependency Inversion Principle


```python
# 5. Dependency Inversion Principle
# Depend on abstractions not on concrete implementation
from abc import ABC,abstractmethod
class Database(ABC):
    @abstractmethod
    def query(self,sql):
        pass
class Mysqldatabase(Database):
    def query(self,sql):
        return f"MySql: {sql}"
class Postgresdatabase(Database):
    def query(self,sql):
        return f"Postgres: {sql}"
class ReportGenerator:
    def __init__(self,db:Database):
        self.db=db
    def generate(self,query):
        return self.db.query(query)
r1=(ReportGenerator(Mysqldatabase()))
r2=(ReportGenerator(Postgresdatabase()))
print(r1.generate("select *"))      
print(r2.generate("select *"))      
```
