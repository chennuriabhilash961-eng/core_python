```python
# 2 A function inside a function(Closure)
def outer():
    message="I am the outer function"
    def inner():
        print(message)
    inner()
outer()
```

```python
# with Parameters
def gr(name):
    print("My name is ",name)
def decorator(func):
    def wrapper():
        print("hi")
        func("Abhi")
    return wrapper
modify1=decorator(gr)
modify1()
```

```python
#1. Create a function start_system()
#     Write a decorator that prints:
#     * “System starting…” before execution
#     * “System started successfully” after execution
def start_system():
    print("System started successfully")
def deco(func):
    def wra():
        print("System Starting")
        func()
    return wra
s=deco(start_system)
s()
```

```python
# 2.     Create a function show_message()
#     Write a decorator that prints:
#     * “Welcome!” before
#     * “Goodbye!” after
def show_message():
    print("Goodbye!")
def dec(func):
    def wrap():
        print("Welcome!")
        func()
    return wrap
mes=dec(show_message)
mes()
```

```python
# 3.     Create a function make_payment()
#     Write a decorator that prints:
#     * “Payment initiated”
#     * “Payment successful”
def make_payment():
    print("Payment Successful")
def dec(func):
    def wrap():
        print("Payment initiated")
        func()
    return wrap
pay=dec(make_payment)
pay()
```
