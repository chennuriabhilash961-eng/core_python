```python
# 3 Given a list of integers, filter numbers divisible by both 2 and 5, add 5 to each using map(), then find the product using reduce().
from functools import reduce
l=[5,10,15,20,25,30]
l1=list(filter(lambda x:x%2==0 and x%5==0,l))
print(l1)
l2=list(map(lambda x:x+5,l1))
print(l2)
l3=reduce(lambda x,y:x*y,l2)
print(l3)
```

```python
# 4. Given a list of integers, write a program to:

# * Filter numbers divisible by 2 but not by 4
# * Add 3 to each using map()
# * Sort the result in descending order
# * Find the product of all elements using reduce()
from functools import reduce
l=[2,4,8,16,22,34]
l1=list(filter(lambda x:x%2==0 and x%4!=0,l))
print(l1)
l2=list(map(lambda x:x+3,l1))
print(l2)
l3=sorted(l2,key=lambda x:x,reverse=True)
print(l3)
l4=reduce(lambda x,y:x*y,l3)
print(l4)
```

```python
# 5. Given a list of words:

# * Filter words that start and end with the same letter
# * Convert them to lowercase
# * Sort by last character, then length
# * Join all words into a single string using reduce()
l=["madam","hello","nan","hi","window","sams"]
l1=list(filter(lambda x:x[0]==x[-1],l))
print(l1)
l2=list(map(lambda x:x.lower(),l1))
print(l2)
l3=sorted(l2,key=lambda x:x[-1])
print(l3)
l4=sorted(l3,key=lambda x:len(x))
print(l4)
l5=reduce(lambda x,y:x+' '+y,l4)
print(l5)
```
