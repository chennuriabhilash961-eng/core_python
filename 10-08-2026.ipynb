```python
def dec1(func):
    def wrapper1(*args,**kwargs):
        print("Hi user")
        func(*args,**kwargs)
        print("applied discount")
    return wrapper1
def dec2(func):
    def wrapper2(*args,**kwargs):
        func(*args,**kwargs)
        print("Applied discount")
    return wrapper2
@dec2
@dec1
def apply_discount(price):
    discount=100
    price-=discount
    print(price)
apply_discount(1000)
```

```python
import functools
def dec1(func):
    @functools.wraps(func)
    def wrapper1(*args,**kwargs):
        print("Hi")
        func(*args,**kwargs)
        print("applied discount")
    return wrapper1
def dec2(func):
    @functools.wraps(func)
    def wrapper2(*args,**kwargs):
        func(*args,**kwargs)
        print("Applied Discount")
    return wrapper2
@dec2
@dec1
def discount(price):
    discount=50
    price-=discount
    print(price)
discount(1000)
print(discount.__name__)
```
