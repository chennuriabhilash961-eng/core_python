```python
# A company wants to generate basic salary information when employee objects are created.
#  Create a class Employee with class variables company = "TechCorp" and employee_count = 0.
#  The constructor should accept name, department, salary, and experience. 
# Validate that salary and experience are not negative. Based on experience,
#  calculate a bonus inside the constructor: employees with more than 5 years receive 15%, employees with 3–5 years receive 10%, and employees with less than 3 years receive 5%. 
# Create an instance dictionary pay_details containing the employee’s name, salary, experience, bonus, and final salary. 
# Generate an employee ID using employee_count. Create three employee objects and display their _dict_.
class Employee:
    company="TechCorp"
    employee_count=0
    def __init__(self,name,department,salary,experience):
        self.name=name
        self.department=department
        if(salary<0 or experience<0):
            print("Invalid salary or experience")
        else:
            self.salary=salary
            self.experience=experience
            Employee.employee_count+=1
            self.employee_id=Employee.employee_count
            if(experience>5):
                bonus=salary*0.15
            elif(experience>=3 and experience<=5):
                bonus=salary*0.1
            else:
                bonus=salary*0.05
            final_salary=salary+bonus
            self.Pay_details={
                'name':name,
                'salary':salary,
                "Experience":experience,
                "bonus":bonus,
                "final_salary":final_salary
            }
emp1=Employee("Anish","It",50000,-5)
emp2=Employee("vennela","Sap",25000,3)
emp3=Employee("hashim","sales",70000,7)

print(emp1.__dict__)
print(emp2.__dict__)
print(emp3.__dict__)
```

```python
#  A mobile store creates a purchase object whenever a customer buys a phone. 
# Create a class MobilePurchase with a class variable store_name = "Smart Mobiles" and purchase_count = 0.
#  The constructor should accept customer, brand, price, storage, and quantity.
#  Validate that price and quantity are positive and that storage is either 64, 128, 256, or 512 GB. Calculate the total price inside the constructor. 
# If the total exceeds ₹50,000, apply a 10% discount; otherwise, apply a 5% discount.
#  Store the complete purchase information in a dictionary called purchase_details. 
# Increment purchase_count for every valid purchase. Create three objects and display their _dict_.
class MobilePurchase:
    store_name="Smart Mobiles"
    purchase_count=0
    def __init__(self,customer,brand,price,storage,quantity):
        self.customer=customer
        self.brand=brand
        if(price>=0 and quantity>=0):
            if(storage==64 or storage==128 or storage==256 or storage==512):
                self.price=price
                self.quantity=quantity
                total=price*quantity
                self.storage=storage
                MobilePurchase.purchase_count+=1
            if(total>50000):
                discount=total*0.10
            else:
                discount=total*0.05
            final_price=total+discount
            self.purchase_details={
                'customer':customer,
                'brand':brand,
                'price':price,
                'quantity':quantity,
                'storage':storage,
                'discount':discount,
                'final_price':final_price
            }
        else:
            print("Invalid")
obj1=MobilePurchase("abhi","iqoo",25000,256,2)
obj2=MobilePurchase("shravan","redmi",10000,128,4)
obj3=MobilePurchase("Ramu","oneplus",35000,512,2)
print(obj1.__dict__)
print(obj2.__dict__)
print(obj3.__dict__)
print(MobilePurchase.purchase_count)
```
