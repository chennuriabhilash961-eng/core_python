```python
class Enrollment:

    # Class variables
    platform = "LearnOnline"
    enrollment_count = 0

    def __init__(self, student_name, course, age):

        # Instance variables
        self.student_name = student_name
        self.course = course
        self.age = age

        # Check age
        if age >= 18:
            self.status = "Registered"
            Enrollment.enrollment_count += 1
        else:
            self.status = "Rejected"


# Creating three objects
s1 = Enrollment("Abhi", "Python", 21)
s2 = Enrollment("Naveen", "Java", 17)
s3 = Enrollment("Rahul", "Data Science", 20)


# Display details
print(s1.student_name, s1.course, s1.age, s1.status)
print(s2.student_name, s2.course, s2.age, s2.status)
print(s3.student_name, s3.course, s3.age, s3.status)


# Display class variables
print("Platform:", Enrollment.platform)
print("Total Registered:", Enrollment.enrollment_count)


# isinstance() check
print("s1 is Enrollment:", isinstance(s1, Enrollment))
print("s2 is Enrollment:", isinstance(s2, Enrollment))
print("s3 is Enrollment:", isinstance(s3, Enrollment))
```
