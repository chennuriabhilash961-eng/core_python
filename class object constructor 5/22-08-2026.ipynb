```python
# Class is an blueprint of an object
# we can create class by using class keyword
# A Class is like an object constructor, or a "blueprint" for creating objects.
# Syntax:
#         class   <classname>:
#                   -------
#                   -------
class classname:
    pass # code of the class block
# we can create custom functions,codes
```

```python
# Object creation is the heart of class execution 
# object creation is equal to function calling
class A:
    pass
obj1=A()
print(type(obj1))
```

```python
obj1=A()
obj2=A()
print(id(obj1))
print(id(obj2))
# both objects location address is different because we aree calling A() (not assigning A(class) to a variable/object) so new objects will created in new memory location 

# m1 is a function f1,f2 are variables to used in function referencing
# In function f1=m1,f2=m1 are has same memory location because we are assigning m1 to f1 and f2
```

```python
# object creation
# __init__ and __new__ these both are helpful in object creation
# init is aconstructor and also a method

# - __new__ method is called first in class
# -new memory space/location is created and allocated returns to __init__
# -__init__ loads/processes properties instance

# we have 2 types of variables in class
# 1.Class variables 
# - These variables are class depended or related to class variables
# - the variables which are belong to class and not depend on object
class A:
    a=20 # class property/variable
obj=A()
print(A.a)
print(obj.a)

# 2.Instance variable
# To use current object we use self
# init's first parameter is object address(self)
class Phone:
    software="Android"
    count=0
    def __init__(self,brand,price):
        self.brand=brand
        self.price=price
        Phone.count+=1
phone1=Phone("Samsung",30000)
phone2=Phone("Vivo",20000)
print(phone1.software)
print(phone1.brand)
print(phone1.price)
print()
print(phone2.software)
print(phone2.price)
print(phone2.brand)
print(Phone.count)

print(phone1.__dict__)
```
