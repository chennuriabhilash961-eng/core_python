```python
# class A:
#     x=20
#     def __init__(self,y):
#         self.y=y
# obj=A()
# TypeError: A.__init__() missing 1 required positional argument: 'y'
```

```python
class A:
    x=20
    def __init__(self,y):
        self.y=y
obj1=A("hey")
obj2=A("hello")
obj3=A("Hi")
# class variable are same to every object but instance variables may vary to every new object
print("Data in obj1")
print(obj1.x)
print(obj1.y)

print("Data in obj2")
print(obj2.x)
print(obj2.y)

print("Data in obj3")
print(obj3.x)
print(obj3.y)

```

```python
# class variable modification 
# syntax: class_name.class_variable=New value
A.x=40
print(obj1.x)
print(obj2.x)
print(obj3.x)

# if we modify class_variable using object it (only changed) create new instance variable in that object not in whole class.
# By this new variable is created in object,but x variable in class wont changes.
obj1.x=100
print(obj1.x)
print(obj2.x)
print(obj3.x)
# firstly it searched in the instance/object variable (local variable ) if it is not there in instance variable then we move to the class variable(global variable) if it is there in class variable then it returns the class variable value
```

```python
#

class A:
    x=20
    def __init__(self,y):
        self.y=y
obj1=A("hey")
obj2=A("hello")
obj3=A("hi")
print(A.__dict__)
print(obj1.__dict__)
print(obj2.__dict__)
# modify
obj1.x = 100  #These changes and updated new values are happens only in which object we are modifying but not in class variable or whole class or in other objects
obj1.z=200
print(obj1.__dict__)
print(obj2.__dict__)
print(obj3.__dict__)
obj2.__dict__['z']=40
print(obj2.__dict__)
# object dictionary variables are instance variables and if we modify in class variable using object then new variables(instance variables) are created.
# by this class variable will not modified but new objects/instance variables will be created that new object will be shadowing

# shadowing: shadowing is done only to particular new instance variable that when we modify  or update 

```
