```python
# 5
# Method Overriding
# Build a payment system with a base class Payment(amount) and three subclasses:
# CreditCard, UPI, and NetBanking. Each overrides a method process() with its own
# logic. Write a function checkout(payment) that calls process() on any payment object
# and demonstrate polymorphism.
class payment:
    def __init__(self,balance):
        self.balance=balance
        print("Object created your current balancee is ",balance)
    def process(self,amount):
        print("processing payment ")
        self.balance+=amount
        print("Your current balance is ",self.balance)
class creditcard(payment):
    def process(self,amount):           
        print("processing payment of Rs.",amount,"via netbanking")
        self.balance+=amount
        print("Your current balance is ",self.balance)
class upi(payment):
    def process(self,amount):
        print("processing payment of Rs.",amount,"via upi")
        self.balance+=amount
        print("Your current balance is ",self.balance)
class netbanking(payment):
    def process(self,amount):
        print("processing payment of Rs.",amount,"via netbanking")
        self.balance+=amount
        print("Your current balance is ",self.balance)
def checkout(payment):
    payment.process(5000)
checkout(upi(2000))

```

```python
# 6
# Duck Typing
# Create three completely unrelated classes: PDFReport, ExcelSheet, and
# EmailMessage. Each has a method send(). Write a function dispatch(item) that calls
# send() using duck typing. Prove that no inheritance is needed.
class pdfreport:
    def send(self):
        print("send pddfreport ...")
class excelsheet:
    def send(self):
        print("send excelsheet ....")
class emailmessage:
    def send(self):
        print("send emailmessage ...")
def dispatch(item):
    item.send()
dispatch(pdfreport())
dispatch(excelsheet())
dispatch(emailmessage())
```

```python
# 7
# Polymorphism + Duck Typing
# Write a function total_area(shapes) that takes a list of any shape objects and returns
# their combined area. Use duck typing - any object with an area() method should
# work. Test with Circle, Rectangle, Triangle, and a custom class Hexagon.
import math
class circle:
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return math.pi*self.radius*self.radius
class rectangle:
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def area(self):
        return self.length*self.breadth
class triangle:
    def __init__(self,base,height):
        self.base=base
        self.height=height
    def area(self):
        return  0.5*self.base*self.height
class hexagon:
    def __init__(self,side):
        self.side=side
    def area(self):
        return (3*math.sqrt(3)/2)*self.side**2
def total_area(shapes):
    area=0
    for shape in shapes:
        area+=shape.area()
    print(area)
shapes=[circle(5),rectangle(10,2),triangle(5,2),hexagon(10)]
total_area(shapes)    
```

```python
# 8
# Concept Question
# Explain the difference between method overriding and method overloading. Python
# does not natively support overloading - show how to simulate it using *args or
# default parameters.

# Main Difference
# Method Overriding	                Method Overloading
# Parent and child classes	        Usually same class
# Requires inheritance	            Does not require inheritance
# Same method name	                Same method name
# Child changes parent's method 	Different arguments are handled
# Example: sound()	                Example: add()
# Supported directly in Python	    Not supported directly in Python
# Uses inheritance	                Simulated using *args or default arguments
```

```python
# 1. Method Overriding

# Method overriding happens when a child class provides its own version of a method that already exists in the parent class.
class Animal:
    def sound(self):
        print("Animal makes sound")
class Dog(Animal):
    def sound(self):
        print("Dog barks")
a = Dog()
a.sound()

# 2. Method Overloading
# Method overloading means having the same method name but different arguments.
class Demo:
    def add(self, a, b):
        print(a + b)
    def add(self, a, b, c):
        print(a + b + c)
# Python will keep only the last add() method.
# So we use *args or default parameters to simulate overloading.

# 3. Simulating Overloading using *args
# *args allows us to pass any number of values.
class Calculator:

    def add(self, *numbers):
        total = 0

        for num in numbers:
            total = total + num

        print("Total:", total)


c = Calculator()

c.add(10, 20)
c.add(10, 20, 30)
c.add(10, 20, 30, 40)
# Method overriding means a child class provides a new implementation of a parent class method, while method overloading means using the same method name with different arguments. Python does not support true method overloading, so we simulate it using *args or default parameters.
```

```python
l = [10,20,30,40]
s = 30
for i in range(len(l)):
    for j in range(i,len(l)):
        sum = 0
        for k in range(i,j+1):
            # print(l[k],end=" ")
            sum += l[k]
        if(sum==s):
            # print(l[k],end=" ")
            print(*l[i:j+1],"->",sum)
```
