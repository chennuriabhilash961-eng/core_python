# Polymorphism

```python
# In python polymorphism means the same function,operation,method  can behave differntly based on the object or data it is working with.
# The word "polymorphism" means "many forms", and in programming it refers to methods/functions/operators with the same name that can be executed on many objects or classes.
# Polymorphism in Python means "many forms". It is a core concept of Object-Oriented Programming (OOP) that allows different objects, functions, or operators to share the same name but behave differently based on their context or data type 
```

```python
class A:
    def m1(self):
        print("m1-A")
class B:
    def m1(self):
        print("m1-B")
class C:
    def m1(self):
        print("m1-C")
a=A()
b=B()
c=C()
a.m1()
b.m1()
c.m1()
```

```python
# Method overriding
# Method overriding happens when a child class provides its own implementation of a method that already exist in its parent class
# the child version replaces the parent version when called on a child object
```

```python
class bank:
    def deposit(self,amount):
        print("Deposited:",amount)
class gpay(bank):
    def deposit(self,amount):
        print("Deposited:",amount)
        print("You have gotten",(amount*0.02),"cashback")
class phonepe(bank):
    def deposit(self,amount):
        print("Deposited:",amount)
        print("You have gotten",(amount*0.01),"cashback")
obj=gpay()
obj.deposit(2000)
obj1=phonepe()
obj1.deposit(2000)
obj3=bank()
obj3.deposit(5000)
```

```python
# Method overloading is not posiible in python 
# example
class A:
    def add(a,b):
        print(a+b)
    def add(a,b,c):
        print(a+b+c)
A.add(10,20)
# python doesn't support traditional method overloading where multiple methods have the same name in the same class
# in the above example the second add() method replaces the first one and if you call A().add(20,30)  error
```

# Duck typing

```python
# Duck typing
# the concept of ducktyping comes from the phrase
# "If it walks like a duck and quacks 
# like a duck then it must be a duck"
# duck typing is python approach to polymorphism without inheritance
# if an object has methods and attributes you need you can use it regardless of its class
# you don't check is a relationship you just call the method if it works great else you get an attribute error
```
