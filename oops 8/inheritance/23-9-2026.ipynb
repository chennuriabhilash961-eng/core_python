```python
# Multi-level inheritance
class A:
    def __init__(self):
        print("A constructor")
class B(A):
    def __init__(self):
        print("B constructor")
        super().__init__()
class C(B):
    def __init__(self):
        print("C constructor")
        super().__init__()
c=C()
print(C.mro())
print(B.mro())
print(A.mro())
```

```python
# Multiple Inheritance
class A:
    def __init__(self):
        print("A constructor")
        super().__init__()
class B:
    def __init__(self):
        print("B constructor")
class C(A,B):
    def __init__(self):
        print("C constructor")
        super().__init__()
obj=C()
print(C.mro())
print(A.mro())
print(B.mro())
print(C.__mro__)
```

```python
# Hierarchical inheritance
class A:
    def __init__(self):
        print("A constructor")
class B(A):
    def __init__(self):
        print("B constructor")
        super().__init__()
class C(A):
    def __init__(self):
        print("C constructor")
        super().__init__()
obj=C()
obj=B()
```

```python
# Hybrid Inheritance
class A:
    def __init__(self):
        print("A constructor")
class B(A):
    def __init__(self):
        print("B constructor")
        super().__init__()
class C(A):
    def __init__(self):
        print("C constructor")
        super().__init__()
class D(B,C):
    def __init__(self):
        print("D constructor")
        super().__init__()
obj=D()
d=D()
y=(x.__name__ for x in D.mro())
for i in y:
    print(i)
```

```python
# Constructor chaining and super chaining
class A:
    def __init__(self):
        print("A constructor")
    def m1(self):
        print("M1-A")
class B(A):
    def __init__(self):
        print("B constructor")
        super().__init__()
    def m1(self):
        print("M1-B")
        super().m1()  
class C(B):
    def __init__(self):
        print("C constructor")
        super().__init__()
    def m1(self):
        print("M1-C")
        super().m1()  
c=C()
c.m1()
```

```python
# create a class employee with instance variable name and salary
# create a child class manager that has another variable called team size
# Initialize variable while creating the object of the manager class
class Employee:
    def __init__(self,name,salary):
        self.name=name
        self.salary=salary
class Manager(Employee):
    def __init__(self,name,salary,team_size):
        self.team_size=team_size
        super().__init__(name,salary)
    def display(self):
        print("Emp name:",self.name)
        print("salary:",self.salary)
        print("Team size:",self.team_size)
obj=Manager("Arun",5000,2)
obj.display()
```
