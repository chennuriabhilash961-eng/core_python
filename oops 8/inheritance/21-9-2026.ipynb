# Inheritance

```python
# Acquiring properties from one class to another class
# Types of inheritance
# 1.Single-level inheritance
# 2.Multi-level inheritance
# 3.Multiple inheritance
# 4.Hierarchical inheritance
# 5.Hybrid Inheritance
```

```python
# 1.single-level inheritance
# Single parent class and single child class
# super/base/parent class
# sub/derived/child class
# B is derived from A

class A:
    def m1(self):
        print("This is m1")
class B(A):
    pass
obj=B()
obj.m1()
```

```python
# 2.Multi-level Inheritance
# One parent class and one child class but b/w these intermediate class
class A:
    def m1(self):
        print("This is m1")
class B(A):
    def m2(self):
        print("This is m2")
class C(B):
    pass
obj=C()
obj.m1()
```

```python
# 3.Multiple Inheritance
# one child class multiple parent class
# one child class takes properties from multiple parent class
# Multiple parent class and single child class
class A:
    def m1(self):
        print("this is m1")
class B:
    def m2(self):
        print("this is m2")
class C(A,B):
    pass
obj=C()
obj.m2()
```

```python
# 4.Hierarchical Inheritance
# Single parent properties multiple child class
class A:
    def m1(self):
        print("This is m1")
class B(A):
    def m2(self):
        print("this is m2")
class C(A):
    def m3(self):
        print("This is m3")
class D(A):
    def m4(self):
        print("This is m4")
obj=D()
obj.m1()
```

```python
# 5.Hybrid Inheritance
# Combination of more than 2 inheritance
class A:
    def m1(self):
        print("this is m1")
class B(A):
    def m2(self):
        print("this is m2")
class C(A):
    def m3(self):
        print("this is m3")
class D(B,C):
    def m4(self):
        print("this is m4")
obj=D()
obj.m1()
```

```python
# 4. Food Ordering System Using Multilevel Inheritance 
# Class 1: Restaurant • Create a method menu(item) that returns the price of the selected food item. 
#  Class 2: FoodCourt (inherits Restaurant) Create the following methods: • display_menu() – Display the available food items. 
#  • order() – Accept the food item from the user and allow multiple orders.  • 
# billing() – Display the total bill and add a packing charge of ₹20. 
#  Class 3: Customer (inherits FoodCourt) • Create an object of the Customer class. 
#  • Call the order() method.  
class Restaurant:
    def menu(self,item):
        if item=='pizza':
            return 150
        elif item=='chicken biryani':
            return 230
        elif item=='Mutton biryani':
            return 600
        else:
            return 0
class Foodcourt(Restaurant):
    def __init__(self):
        self.total=0
    def display_menu(self):
        print("Pizza - 150")
        print("Chicken Biryani - 230")
        print("Mutton Biryani - 600")
    def order(self):
        while True:
            item=input("Enter your item:")
            price=self.menu(item)
            if(price==0):
                print("Item not available")
            else:
                self.total+=price
                print("Added:",item,price)
            choice=input("Do you want to add more (yes/no):")
            if(choice=='no'):
                break
    def billing(self):
        print("Billing Details")
        print("----------------")
        print("Cost of Food",self.total)
        print("Packing charges:20")
        self.total+=20
        print("Total bill:",self.total)
class Customer(Foodcourt):
    def __init__(self):
        super().__init__()
obj=Customer()
obj.display_menu()
obj.order()
obj.billing()
```

```python
# 5. Movie Ticket Booking System Using Multilevel Inheritance 
# Class 1: Movie 
# • Create a method ticket(movie) that returns the ticket price.  
# Class 2: Booking (inherits Movie) 
# Create the following methods: 
# • movies() – Display the available movies.  
# • selection() – Allow the user to book multiple tickets.  
# • billing() – Display the total amount and add a booking charge of ₹30.  
# Class 3: Customer (inherits Booking) 
# • Create an object and call the selection() method. 
class Movie:
    def ticket(self,movie):
        if movie=="Salaar":
            return 350
        elif movie=='Paradise':
            return 200
        elif movie=="varanasi":
            return 500
        else:
            return 0
class Booking(Movie):
    def __init__(self):
        self.total=0
    def movies(self):
        print("1.Salaar - 350")
        print("2.Paradise - 200")
        print("3.Varanasi - 500")
    def selection(self):
        while True:
            movie=input("Enter your movie:")
            price=self.ticket(movie)
            if price==0:
                print("Movie is not available")
            else:
                self.total+=price
                print("Movie Added:",movie,price)
            choice=input("Do you want to book tickets for more movies (yes/no):")
            if(choice=='no'):
                break
    def billing(self):
        print("Booking Details")
        print("---------------")
        print("Total price",self.total)
        print("Booking charge:30")
        self.total+=30
        print("Final price",self.total)
class Customer(Booking):
    def __init__(self):
        super().__init__()
m=Customer()
m.movies()
m.selection()
m.billing()


    
```

```python
# 9. Bus Ticket Booking System Using Hierarchical Inheritance 
# Class 1: Bus 
# • Create methods to return the fare for Sleeper, Semi-Sleeper, and AC 
# buses.  
# Class 2: RedBus (inherits Bus) 
# • Create the methods routes(), booking(), and billing().  
# • Add 10% GST and a reservation charge of ₹30.  
# Class 3: AbhiBus (inherits Bus) 
# • Create the methods routes(), booking(), and billing().  
# • Add 10% GST and a reservation charge of ₹20.  
# Driver Code 
# • Ask the user to choose the platform and call the booking() method. 
class Bus:
    def buses(self,fare):
        if fare=='Sleeper':
            return 500
        elif fare=='Semi-Sleeper':
            return 300
        elif fare=='Ac':
            return 800
class RedBus(Bus):
    def __init__(self):
        self.price=0
        super().__init__()
    def routes(self):
        print("1.Hyderabad -> Vizag")
        print("2.Vizag -> Hyderabad")
        print("3.Warangal -> goa")
        print("4.goa -> warangal")
    def booking(self):
        fare=input("Enter your coaching")
        self.price=self.buses(fare)
        print("Total price:",self.price)
    def billing(self):
        print("Billing --- Details")
        gst=0.10*self.price
        self.price+=gst
        print("After adding gst of 10%",self.price)
        print("Reseravation charge 30Rs")
        self.price+=30
        print("Final price",self.price)
class AbhiBus(Bus):
    def __init__(self):
        self.price=0
        super().__init__()
    def routes(self):
        print("1.Hyderabad -> Vizag")
        print("2.Vizag -> Hyderabad")
        print("3.Warangal -> goa")
        print("4.goa -> warangal")
    def booking(self):
        fare=input("Enter your coach")
        self.price=self.buses(fare)
        print("Total price",self.price)
    def billing(self):
        print("Billing Details")
        gst=0.10*self.price
        self.price+=gst
        print("Added gst of 10%:",self.price)
        print("Reservation charge is 20RS")
        self.price+=20
        print("Final booking price is:",self.price)
choice=input("(RedBus/AbhiBus)")
if choice=='AbhiBus':
    obj=AbhiBus()
    obj.routes()
    obj.booking()
    obj.billing()
else:
    obj=RedBus()
    obj.routes()
    obj.booking()
    obj.billing()
```

```python
# 10. ATM System Using Multiple Inheritance 
# Class 1: SBI 
# • Create the methods deposit(amount) and check_balance().  
# Class 2: UnionBank 
# • Create the methods withdraw(amount) and mini_statement().  
# Class 3: ATM (inherits SBI and UnionBank) 
# • Create the methods menu() and transaction().  
# • Allow the user to perform banking operations.  
# Driver Code 
# • Create an object of the ATM class.  
# • Call the transaction() method.
class sbi:
    def __init__(self,balance):
        self.balance=balance
        self.trans=[]
    def deposit(self,amount):
        if(amount>0):
            self.balance+=amount
            print("Deposited:",amount)
        else:
            print("Invalid amount")
    def checkbalance(self):
        print("Current balance:",self.balance)
class unionbank:
    def withdraw(self,amount):
        if(amount<=self.balance):
            self.balance-=amount
            self.trans.append(amount) 
            print("Withdrawn:",amount)
        else:
            if(amount>self.balance):
                print("Insufficient funds")
            else:
                print("Invalid amount")
    def mini_statement(self):
        print("Mini statement")
        for i in self.trans:
            print(i)
        print("Balance:",self.balance)
class atm(sbi,unionbank):
    def __init__(self, balance):
        super().__init__(balance)
    def menu(self):
        print("----Menu----")
        print("1.withdrawn")
        print("2.Deposit")
        print("3.Checkbalance")
        print("4.Mini statement")
        print("5.Exit")
    def transaction(self):
        self.menu()
        while True:
            choice=int(input("Enter any number:"))
            if choice==1:
                amount=int(input("Enter amount"))
                self.withdraw(amount)
            elif choice==2:
                amount=int(input("Enter amount"))
                self.deposit(amount)
            elif choice==3:
                self.checkbalance()
            elif choice==4:
                self.mini_statement()
            elif choice==5:
                print("Thank you")
                break
obj=atm(100000)
obj.transaction()
```
