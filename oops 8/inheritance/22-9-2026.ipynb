```python
# super()
# It will pass current object address to child class
# super() returns a proxy object that delegates method calls to the next class in MRO(Method Resolution Order)
# It lets you call to the next class property without specifying the name of the parent
class B(A):
    c3="Hello"
    def __init__(self,x,y,z):
        self.z=z
        # A.__init__(self,20,30)
        super().__init__(20,30)
```

```python
# Multi level inheritance
class A:
    def __init__(self,x):
        self.x=x
class B(A):
    def __init__(self,x,y):
        self.y=y
        A.__init__(self,x)
class C(B):
    def __init__(self,x,y,z):
        self.z=z
        B.__init__(self,x,y)

obj=C(20,50,90)
print(obj.x)
print(obj.y)
print(obj.z)
```

```python
# constructor chaining and super chaining
class A:
    def __init__(self):
        print("A consttructor")
    def m1(self):
        print("m1-A")
class B(A):
    def __init__(self):
        print("B consttructor")
        super().__init__()
    def m1(self):
        print("m1-B")
        super().m1()
class C(B):
    def __init__(self):
        print("C consttructor")
        super().__init__()
    def m1(self):
        print("m1-C")
        super().m1()
c=C()
c.m1()
```
