```python
# 1. Payment System 
# Create an abstract class Payment with an abstract method pay(amount)
# Create:
# •	UPI
# •	CreditCard
# •	Cash
# Each class should implement pay() differently.
from abc import ABC,abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
class Upi(Payment):
    def pay(self,amount):
        print("upi payment Rs",amount)
class creditcard(Payment):
    def pay(self,amount):
        print("credit card payment ",amount)
class cash(Payment):
    def pay(self,amount):
        print("cash payment ",amount)
u=Upi()
c=creditcard()
ca=cash()
u.pay(2000)
c.pay(4000)
ca.pay(8000)
```

```python
# 2. Notification System 
# Create an abstract class Notification with:
# send(message)
# Create:
# •	Email 
# •	SMS 
# •	WhatsApp 
# Each child class should implement send() differently.
from abc import ABC,abstractmethod
class Notification(ABC):
    @abstractmethod
    def send(self,message):
        pass
class Email(Notification):
    def send(self,message):
        print("Email mesage:",message)
class Sms(Notification):
    def send(self,message):
        print("Sms mesage:",message)
class Whatsapp(Notification):
    def send(self,message):
        print("Whatsapp mesage:",message)
email=Email()
sms=Sms()
whatsapp=Whatsapp()
email.send("Hey")
sms.send("Hi")
whatsapp.send("Hello")
```

```python
# 3. Login System
# Create an abstract class LoginSystem with:
# login(username, password)
# logout()
# Create:
# •	AdminLogin 
# •	UserLogin 
# Implement login differently for admin and normal users.
from abc import ABC,abstractmethod
class LoginSystem(ABC):
    @abstractmethod
    def login(self,username,password):
        pass
    @abstractmethod
    def logout(self):
        pass
class AdminLogin(LoginSystem):
    def login(self,username,password):
        print("Admin username:",username)
        print("Admin password:",password)
    def logout(self):
        print("Admin logged out")
class UserLogin(LoginSystem):
    def login(self,username,password):
        print("User username:",username)
        print("User password:",password)
    def logout(self):
        print("User logged out")
admin=AdminLogin()
user=UserLogin()
admin.login("Hari",9789)
user.login("David",7898)
```
