# Abstraction

```python
# Hiding all the implementation details and display only required for the user.
# Abstraction is the process of hiding implementation and exposing only essential features
# Abstraction tells us what an object can do without telling us how it doesn't.
# Abstraction is the process of hiding implementation details and exposing only the essential functionality to the user. It is used to hide the implementation details from the user and expose only necessary parts, making the code simpler and easier to interact with.


# Why do we need abstraction
# 1.Hide complexity
# 2.support polymorphism
# 3.provide a common structure
```

Abstract class

```python
# Abstract class is a class designed as a blueprint for related classes.It may contain abstract methods normal methods constructor and variable
# An abstract class containing unimplemented abstract methods cannot be instantiated
# An abstract method is a method declared in the parent class that requires a concrete sub class to provide an implementation.
# It is marked using @abstractmethod decorator
# using super() we can call the constructor in the abstract class and sub class

# concrete class: object can be created using the concrete class
# abstract class:but can't create object in abstract class

# Abstract Base Class
# Abstract Base Class (ABC) is used to achieve data abstraction by defining a common interface for its subclasses. It cannot be instantiated directly and serves as a blueprint for other classes.


# 1. Abstract Method
# Abstract methods are method declarations without a body defined inside an abstract class. They act as placeholders that force subclasses to provide their own specific implementation, ensuring consistent structure across derived classes.

# 2. Concrete Method
# Concrete methods are fully implemented methods within an abstract class. Subclasses can inherit and use them directly, promoting code reuse without needing to redefine common functionality.
```

```python
# 1. Import ABC and abstractmethod from abc.
# 2. An abstract class can have abstract methods, normal methods, constructors and variables.
# 3. An abstract class with unimplemented abstract methods cannot be instantiated.
# 4. A child must implement all inherited abstract methods to become concrete.
# 5. Abstractness can continue through multilevel inheritance.
# 6. Python does not automatically enforce an exact abstract-method signature.
# 7. Changing parameter names is harmless; changing required parameter count can break the expected contract.
# 8. super() can call an abstract parent's constructor or normal method.
# 9. Abstract classes are useful for enforcing a common design while supporting polymorphism.
```

```python
from abc import ABC,abstractmethod
class whatsapp(ABC):
    @abstractmethod
    def send(self,data):
        pass
class sendphoto(whatsapp):
    def send(self):
        print("Compress image")
        print("encrypt image")
        print("decrypt image")
class sendtext(whatsapp):
    def send(self):
        print("checking character limit")
        print("Encrypt message")
        print("sending message")
        print("decrypting message")
        print("message sent !!")
message1=sendphoto()
image1=sendtext()
message1.send()
print()
image1.send()
# It will work if there is no @abstractmethod but it is not correct
# The same method name should be there for every child class
```

```python
# create a python appn with an abstract class and upload with an abstract method  upload content implement three sub classes photo upload ,video upload and story upload
from abc import ABC,abstractmethod
class insta:
    @abstractmethod
    def upload_content(self,data):
        pass
class photoupload(insta):
    def upload_content(self):
        print("choose from gallery")
        print("photo is uploading")
        print("photo is  successfully uploaded")
class videoupload(insta):
    def upload_content(self):
        print("choose from gallery")
        print("video is uploading")
        print("video is  successfully uploaded")
class storyupload(insta):
    def upload_content(self):
        print("choose from gallery")
        print("story is uploading")
        print("story is  successfully uploaded")
ph=photoupload()
vi=videoupload()
st=storyupload()
ph.upload_content()
print()
vi.upload_content()
print()
st.upload_content()
    
```
