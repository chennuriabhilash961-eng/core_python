```python
# 1. Payment System 
# Create an abstract class Payment with an abstract method pay(amount)
# Create:
# •	UPI
# •	CreditCard
# •	Cash
# Each class should implement pay() differently.
from abc import ABC,abstractmethod
class payment(ABC):
    @abstractmethod
    def pay(self,amount):
        pass
class upi(payment):
    def pay(self,amount):
        print("Upi is processing")
        print("upi payment",amount,"is successfully paid")
class creditcard(payment):
    def pay(self,amount):
        print("creditcard is processing")
        print("creditcard payment",amount,"is successfully paid")
class cash(payment):
    def pay(self,amount):
        print("cash is processing")
        print("cash payment",amount,"is successfully paid")
u=upi()
c=creditcard()
ca=cash()
u.pay(2000)
print()
c.pay(3900)
print()
ca.pay(7000)
```

```python
# 2. Notification System 
# Create an abstract class Notification with:
# send(message)
# Create:
# •	Email 
# •	SMS 
# •	WhatsApp 
# Each child class should implement send() differently.
from abc import ABC,abstractmethod
class notification(ABC):
    @abstractmethod
    def send(self,message):
        pass
class Email(notification):
    def send(self,message):
        print("Email message is:",message)
class Sms(notification):
    def send(self,message):
        print("sms message is:",message)
class Whatsapp(notification):
    def send(self,message):
        print("whatsapp message is:",message)
email=Email()
sms=Sms()
whatsapp=Whatsapp()
email.send("HI")
sms.send("Hello")
whatsapp.send("Hey what's up")
```

```python
# 3. Login System
# Create an abstract class LoginSystem with:
# login(username, password)
# logout()
# Create:
# •	AdminLogin 
# •	UserLogin 
# Implement login differently for admin and normal users.
from abc import ABC,abstractmethod
class Loginsystem(ABC):
    @abstractmethod
    def login(self,username,password):
        pass
    @abstractmethod
    def logout(self):
        pass
class AdminLogin(Loginsystem):
    def login(self,username,password):
        print("Admin Username is:",username)
        print("Admin Password is:",password)
    def logout(self):
        print("logged out")
class UserLogin(Loginsystem):
    def login(self,username,password):
        print("User Username is:",username)
        print("User Password is:",password)
    def logout(self):
        print("logged out")
admin=AdminLogin()
user=UserLogin()
admin.login("Abhi",1234)
admin.logout()
user.login("Varun",9876)
```

```python
# 4. Bank Account
# Create an abstract class BankAccount with two abstract methods:
# deposit(amount)
# withdraw(amount)
# Create:
# •	SavingsAccount 
# •	CurrentAccount 
# Maintain the balance using an instance variable.
from abc import ABC,abstractmethod
class bankaccount(ABC):
    @abstractmethod
    def deposit(self,amount):
        pass
    @abstractmethod
    def withdraw(self,amount):
        pass
class SavingsAccount(bankaccount):
    def deposit(self,amount):
        print("savings account Amount deposited:",amount)
    def withdraw(self,amount):
        print("savings account Amount withdrawn:",amount)
class CurrentAccount(bankaccount):
    def deposit(self,amount):
        print("current account Amount deposited:",amount)
    def withdraw(self,amount):
        print("current account Amount withdrawn:",amount)
savings=SavingsAccount()
current=CurrentAccount()
savings.deposit(2000)
savings.withdraw(1000)
current.deposit(1500)
current.withdraw(500)
```

```python
# 14
# Abstract Contract Enforcement
# Create an abstract class Animal(ABC) with abstract methods speak() and move(),
# and a concrete method breathe() shared by all. Implement Dog, Bird, and Fish.
# Prove the contract is enforced by trying to instantiate Animal or a subclass that
# forgot to implement speak().
from abc import ABC,abstractmethod
class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass
    @abstractmethod
    def move(self):
        pass
    def breathe(self):
        print('Animal is breathing')
class Dog(Animal):
    def speak(self):
        print("Dog is barking")
    def move(self):
        print("dog is running")
class Bird(Animal):
    def speak(self):
        print("Bird is chirps")
    def move(self):
        print("Bird is flying")
class Fish(Animal):
    def speak(self):
        print("Fish has no sound")
    def move(self):
        print("Fish is swimming")
dog=Dog()
bird=Bird()
fish=Fish()
dog.speak()
dog.move()
dog.breathe()
print()
bird.speak()
bird.move()
bird.breathe()
print()
fish.speak()
fish.move()
fish.breathe()

```
