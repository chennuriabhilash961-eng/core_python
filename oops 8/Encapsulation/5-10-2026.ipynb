# Encapsulation

```python
# Encapsulation is a process of protecting the data within a class and restricting directing access to data
# Python mainly uses '_' underscores to indicate access levels
# Access levels:
# 1.name=> public access =>no restriction,can be accessed/modified directly
# 2._age=> protected =>only that class & subclasses
# 3.__phone_number=> private =>direct access only in that class # undergoes name mangling 
# objname._classname__variablename by using this we can access the private data
```

```python
# 1. Access Levels
# • Public: Accessible anywhere; regular naming (self.name).
# • Protected: Intended for internal use or subclasses; single underscore prefix (self._balance) by convention without strict enforcement.
# • Private: Intended to be hidden; double underscore prefix (self.__account_number) enforced via Name Mangling, renaming attributes internally to _ClassName__variable_name.
```

```python
# @property

# @property allows us to access a private variable using normal attribute syntax

class Product:
    def __init__(self,price):
        self.__price=price
    @property
    def price(self):
        return self.__price
product1=Product(10000)
print(product1.price)
```

```python
# @property.setter
# @property.setter is used to modify a private variable with modification

class Product:
    def __init__(self,price):
        self.__price=price
    @property
    def price(self):
        return self.__price
    @price.setter
    def price(self,new_price):
        if(new_price>0):
            self.__price=new_price
pr1=Product(5000)
print(pr1.price)
print()
pr1.price=100
print(pr1.price)
```

```python
# 9
# @property + @setter
# Design a BankAccount class with a private_balance. Create @property balance as
# a getter. Create a @balance.setter that raises ValueError if the value is negative.
# Add methods deposit(amount) and withdraw(amount) that update balance through
# the setter. Test all edge cases.
class bankaccount:
    def __init__(self,balance):
        self.__balance=balance
    @property
    def balance(self):
        return self.__balance
    @balance.setter
    def balance(self,new_balance):
        if(new_balance<0):
            raise ValueError("Balance cannot be negative")
        else:
            self.__balance=new_balance
    def deposit(self,amount):
        if(amount>0):
            self.__balance+=amount
        else:
            print("Invalid amount")
    def withdraw(self,amount):
        if(amount<self.__balance):
            self.__balance-=amount
        else:
            print("Insufficient funds")
obj=bankaccount(8000)
print("current balance is:",obj.balance)

obj.deposit(2000)
print("current balance after deposit is:",obj.balance)


obj.withdraw(7000)
print("current balance after withdraw is:",obj.balance)

# obj.balance=-2000
        
```

```python
# 10
# Computed Properties
# Create a class Circle with a private_radius. Add @property radius (getter) and
# @radius.setter that validates the radius is positive. Add a computed @property area
# and @property circumference. Show that changing radius automatically updates
# both computed properties.
import math
class circle:
    def __init__(self,radius):
        self.__radius=radius
    @property
    def radius(self):
        return self.__radius
    @radius.setter
    def radius(self,new_radius):
        if(new_radius>0):
            self.__radius=new_radius
        else:
            print("radius should be positive")
    @property
    def area(self):
        return math.pi*self.__radius*self.__radius
    @property
    def circumference(self):
        return 2*math.pi*self.__radius
obj=circle(5)
print("Radius:",obj.radius)
print("Area:",obj.area)
print("Circumference:",obj.circumference)

obj.radius=10
print("Radius:",obj.radius)
print("Area:",obj.area)
print("Circumference:",obj.circumference)


obj.radius=-10
```

```python

import math
class circle:
    def __init__(self,radius):
        self.__radius=radius
    @property
    def radius(self):
        return self.__radius
    def area(self):
        return math.pi*self.__radius*self.__radius
    def circumference(self):
        return 2*math.pi*self.__radius
    @radius.setter
    def radius(self,new_radius):
        if(new_radius>0):
            self.__radius=new_radius
        else:
            print("radius should be positive")
obj=circle(5)
print("Radius:",obj.radius)
print("Area:",obj.area())
print("Circumference:",obj.circumference())

obj.radius=10
print("Radius:",obj.radius)
print("Area:",obj.area())
print("Circumference:",obj.circumference())


obj.radius=-10

```

```python
# 11
# Access Modifiers
# Write a class Config with protected _settings dict and private_secret_key.
# Demonstrate: (a) public attribute access, (b) protected access with a warning
# comment, (c) private access via name mangling (and explain why you should NOT
# do this in real code), (d) a safe public method get_setting(key) as the correct
# approach.
class config:
    def __init__(self):
        self.app_name="Application name"
        self._settings={
            "theme":"white",
            "language":"English"
        }
        self.__secret_key="ABC123"
    def get_setting(self,key):
        return self._settings.get(key,"setting not found")
obj=config()
print("App name :",obj.app_name)
print("Settings :",obj._settings)
print("Secret_key :",obj._config__secret_key)    # being a name mangling

        
print("theme:",obj.get_setting("theme"))    
print("language:",obj.get_setting("language"))    
print("font:",obj.get_setting("font"))    
    
```

```python
# 12
# Property Chaining
# Build a Person class using @property for first_name, last_name, and full_name.
# full_name should be a computed read-only property. Add a @full_name.setter that
# splits the input string on a space and updates first and last names separately.
class person:
    def __init__(self,first_name,last_name):
        self.__first_name=first_name
        self.__last_name=last_name
    @property
    def first_name(self):
        return self.__first_name
    @property
    def last_name(self):
        return self.__last_name
    @property
    def fullname(self):
        return self.__first_name+ " " +self.__last_name
    @fullname.setter
    def fullname(self,value):
        first,last=value.split(" ")
        self.__first_name=first
        self.__last_name=last
obj=person("Abhilash","Chennuri")
print(obj.first_name)
print(obj.last_name)
print(obj.fullname)
print()
obj.fullname="macha sukumar"
print(obj.first_name)
print(obj.last_name)
print(obj.fullname)
    
```
