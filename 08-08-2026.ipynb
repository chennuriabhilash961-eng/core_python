```python
# Implementing in two ways one is manual and other is shorthand format
def decorator(func):
    def wrapper(*args,**kwargs):
        print("before calling")
        func(*args,**kwargs)
        print("after calling")
    return wrapper
def add(a,b):
    print(a+b)
add=decorator(add)
add(10,20)
```

```python
# shorthand format
def decorator(func):
    def wrapper(*args,**kwargs):
        print("before calling")
        func(*args,**kwargs)
        print("after calling")
    return wrapper
@decorator
def add(a,b):
    print(a+b)
add(20,50)
```

The @decorator syntax

```python
# Python provides a cleaner more readable shorthand for applying decorators the @ symbol placed on the line immediately above the function definition. 
# syntax
# @decorator Syntax
# @my_decorator
# def my_function ():
### This is exactly equivalent to writing
# def my_function():
#     ...
# my_function=my_decorator(my_function)
```

```python
# example
def shout(func):
    def wrapper(*args,**kwargs):
        l=func(*args,**kwargs)
        return str(l).upper()
    return wrapper
@shout
def greet(name):
    return f"hello, {name}"
print(greet("alice"))
```

```python
# 1.     Create a function get_message() that returns "hello user". Write a decorator using @ syntax that converts the output to uppercase.
def decorator(func):
    def wrapper(*args,**kwargs):
        l=func(*args,**kwargs)
        return str(l).upper()
    return wrapper
@decorator
def get_message():
    return f"hello, user"
print(get_message())
```

```python
# 2.     Create a function get_number() that returns 10
#     Use a decorator to return double the value.
def decorator(func):
    def wrapper(*args,**kwargs):
        x=func(*args,**kwargs)
        x=x*2
        return x
    return wrapper
@decorator
def get_number():
    return 10
print(get_number())
```

```python
# 3.     Create a function place_order(item)
#     Use a decorator to print:
#     * “Order process started”
#     * “Order process completed”
def decorator(func):
    def wrapper(*args,**kwargs):
        print("order process started")
        func(*args,**kwargs)
        print("order process completed")
    return wrapper
@decorator
def place_order(item):
    print("Item:",item)
place_order("Pizza")
```

```python
# 4.     Create a function login(username)
#     Use a decorator to print:
#     * “Authenticating user…”
#     * “Login successful”
def decorator(func):
    def wrapper(*args,**kwargs):
        print("Authenticating user")
        func(*args,**kwargs)
        print("Login successful")
    return wrapper
@decorator
def login(username):
    print(username)
login("Abhi")
```

```python
# 5.     Create a function send_message(msg)
#     Use a decorator to print:
#     * “Sending message…”
#     * “Message sent”
def decorator(func):
    def wrapper(*args,**kwargs):
        print("sending message")
        func(*args,**kwargs)
        print("message sent")
    return wrapper
@decorator
def send_message(msg):
    print(msg)
send_message("Hi ,How are you!")
```

```python
# 7.     Create a function apply_discount(price)
#     Use a decorator to print:
#     * “Applying discount…”
#     * “Discount applied”
def decorator(func):
    def wrapper(*args,**kwargs):
        print("Applying discount")
        func(*args,**kwargs)
        print("Discount applied")
    return wrapper
@decorator
def apply_discount(price):
    price=price-(price*0.10)
    print(price)
apply_discount(1000)
```
