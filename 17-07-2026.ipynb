# Lambda Expression

```python
# Lambda expression
# It creates an single line
# lambda x,y:x+y
#There is no return in lambda
# Syntax for condition if else: lambda ______:true expre  if condition else false expre
# Lambda is a small anonymous function return in a single line its useful for short through away applications or operations especially passsing a function as an argument
# we cannot pass loops but we can pass stmts
```

```python
add=lambda x,y:x+y
print(add(5,6))
```

```python
hi=lambda x:"even" if x%2==0 else "odd"
print(hi(6))
print(hi(5))
```

```python
# sorting the elements in order lambda functions
l=[2,5,6,2,6,9,24,14]
l.sort(key=lambda x:x)
print(l)
# it sort in the order division
l=[2,5,6,2,6,9,24,14]
l.sort(key=lambda x:x/2)
print(l)
# it sorted in the order remainder=0,1,2
l=[2,5,6,2,6,9,24,14]
l.sort(key=lambda x:x%2)
print(l)
# it sorted as reverse to the sort
l=[2,5,6,2,6,9,24,14]
l.sort(key=lambda x:-x)
print(l)

```

```python
# In list there are multiple tuples
fruits=[(1,'banana'),(2,'apple'),(3,'cherry')]
fruits.sort(key=lambda x:x[1])
print(fruits)
```

```python
# 1 Write a lambda function that takes a number and returns its cube

square=lambda a:a**3
print(square(5))
```

```python
# 2 create a lambda that takes two numbers and returns the larger one using a conditional expression(x if x>y else y)
larger=lambda x,y:x if x>y else y
print(larger(20,9))
```

```python
# 3 convert this regular function into a lambda: def  even(n): return n%2==0
even=lambda n:"even"  if n%2==0 else "odd"
print(even(6))
print(even(5))
```

```python
# 4 Use a lambda with .sort() to sort this list of tuples by the second element:[(1,'banana'),(2,'apple'),(3,'cherry')]

fruit=[(1,'banana'),(2,'apple'),(3,'cherry')]
fruit.sort(key=lambda x:x[1])
print(fruit)
```

```python
# 5 can a lambda function call another function inside it? write an example
# yes a lambda function call another function inside its expression provided the target function is defined or imported before the lambda is executed
# Example
def  square(x):
    return x*x

squares=lambda x: square(x)
print(squares(5))

add=lambda x,y:x+y
multiply=lambda a,b:a*b+add(a,b)
print(multiply(10,20))
```

```python
# 6 what are the three main limitations of lambda compared to def?List them
# Single Expression Only: Lambda functions can only contain a single expression and cannot include statements (such as if blocks, for loops, try-except blocks, or multiple lines of code). 
# Anonymity and Debugging: Lambdas are unnamed (anonymous), meaning they lack a descriptive name which makes debugging more difficult and prevents the use of docstrings or type hints for documentation. 
# Limited Reusability: Because they are designed for short, throwaway operations, lambdas have limited reusability and are generally unsuitable for complex logic that requires maintenance or multiple operations.
# no return statement in lambda function
```

```python
def iseven(x):
    if(x%2==0):
        return True
    else:
        return False
    
l=[17,29,14,9,6,4]
l1=list(filter(iseven,l))
print(l1)

l2=list(filter(lambda x:x%2==0,l))
print(l2)

l3=[1,2,3,4]
p=list(map(lambda x:x**2,l3))
print(p)

print(list(filter(iseven,p)))

print(list(filter(lambda x:x%2==0,p)))
print(list(filter(lambda x:x%2==0,list(map(lambda x:x**2,l3)))))
```
