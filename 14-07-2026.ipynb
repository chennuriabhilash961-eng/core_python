```python
# shallow copy here the l value changed and sc also changed
l=[[10,20],[30,40]]
import copy
sc=copy.copy(l)
sc[0][1]=25
print(sc)
print(l)
```

```python
#deepcopy here before deep copy l[] is same after also same l[] but it is changed dc
l=[[10,20],[30,40]]
import copy
dc=copy.deepcopy(l)
dc[0][1]=25
print(sc)
print(l)
```
