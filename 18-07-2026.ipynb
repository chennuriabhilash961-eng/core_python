```python
#Higher order functions
# A Higher order function if any function that takes another function as an argument returns a function
# Python has four built in higher order function 
# 1.Map-Apply a function on every element of an iterable
# 2.Filter-keep only elements where a function returns True
# 3.Reduce-combine all elements into a single value
# 4.sorted-sort any iterable using a custom function
```

# 1.Map

```python
# map()
# syntax:map(function,iterable)
l=[1,2,3,4]
l1=list(map(lambda x:x**2,l))
print(l1)
```

```python
l1=[1,2,3,4]
l2=[5,6,7,8]
l3=list(map(lambda a,b: a+b,l1,l2))
print(l3)
```

```python
s=[3,7,9,11]
f=list(map(lambda x:x/2,s))
print(f)

s=[3,7,9,11]
f=list(map(lambda x:x//2,s))
print(f)
```

# Filter

```python
def iseven(x):
    if(x%2==0):
        return True
    else:
        return False
    
l=[16,43,54,7,8,1]
l1=list(filter(iseven,l))
print(l1)
print(list(filter(lambda x:x%2==0,l)))
```

```python
l=[1,2,3,4]
p=list(map(lambda x:x**2,l))
print(p)
print(list(filter(lambda x:x%2==0,l)))
print(list(filter(lambda x:x%2==0,p)))
print(list(filter(lambda x:x%2==0,list(map(lambda x:x**2,p)))))

```

```python
# - A list of usernames is stored in lowercase. Use map() to format them so that the first letter is uppercase.
username=["abhi","vamsshi","hari","pashi"]
title=list(map(lambda a:a.title(),username))
print(title)

username=["abhi","vamsshi","hari","pashi"]
title=list(map(lambda a:a.capitalize(),username))
print(title)

```

```python
# - An e-commerce website wants to display only products priced above ₹500. Use filter() to extract those prices from a list.
l=[550,45,389,509,23,1000]
l2=list(filter(lambda x:x if x>=500 else "",l))
print(l2)

l=[550,45,389,520,23,1000]
l2=list(filter(lambda x:x>500 ,l))
print(l2)
```

```python
# - Use map() with a lambda function to multiply each number in a list by 5.
l=[2,3,4,5,6]
l1=list(map(lambda x:x*5,l))
print(l1)
```

```python
# - Write a program that uses map() to calculate the length of each word in a list of strings.
l=["abhhhdhi","ahdrun","awdbwdbjay"]
l2=list(map(len,l))
print(l2)

l=["abhhhdhi","ahdrun","awdbwdbjay"]
l2=list(map(lambda x:len(x),l))
print(l2)
```

```python
# - Given a list of integers, use filter() to select numbers greater than 50.
l=[85,-1,95,-43,2,-5]
l2=list(filter(lambda x:"x" if x>50 else "",l))
print(l2)

l=[85,-1,95,-43,2,-5]
l2=list(filter(lambda x:x>50,l))
print(l2)
```

```python
# - Use filter() with a lambda function to select numbers that are multiples of 4.
l=[4,6,16,27,32,55,64]
l2=list(filter(lambda x:x%4==0,l))
print(l2)
```
