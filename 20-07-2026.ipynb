```python
l=[10,20,3,7,4,9]
l1=list(map(lambda x:x**2,l))
l2=list(filter(lambda x:x%4==0,l1))
print(l2)
# single pipeline solution calling in asingle line
print(list(filter(lambda x:x%4==0,list(map(lambda x:x**2,l)))))
```

# Reduce

```python
# Reduce:reduce()
# it is used for single value
# syntax:reduce(function,iterable)
# Reduce applies a function cumulatively the result of each step is pass to the next as argument(first argument)
# max,min avg,prod
# for reduce we should import reduce
# from functools import reduce
```

```python
from functools import reduce
l=[2,3,47,9,14]
print(reduce(lambda x,y:x+y,l))
```

```python
# for max value
from functools import reduce
l=[1,43,13,41,9,5,67]
print(reduce(lambda x,y:x if x>y else y,l))
# for min value
print(reduce(lambda x,y:x if x<y else y,l))
```

```python
# concatenate of list of strings
st=['hi','hello','bye']
print(reduce(lambda x,y:x+y,st))
print(reduce(lambda x,y:x+' '+y,st))
print(reduce(lambda x,y:x+'@'+y,st))
```

```python
# 1 Given a list of strings, write a program using reduce() to concatenate all strings into a single string.
l=["abhi","hari","vamshi"]
print(reduce(lambda x,y:x+' '+y,l))
```

```python
# 2 Given a list of digits, write a program using reduce() to form a single number (e.g., [1,2,3] → 123).
# l=[1,2,3,4,5]
from functools import reduce

digits = [1, 2, 3]
number = reduce(lambda x, y: x * 10 + y, digits)
print(number)  # Output: 123  
digits = [1, 2, 3]
number=reduce(lambda x,y:str(x)+''+str(y),digits)
print(number)
```

```python
# 3. ⁠Given a list of numbers, write a program using reduce() to calculate the cumulative difference 
l=[2,345,44,78,3,2,5]
print(reduce(lambda x,y:x-y,l))

```

```python
# ⁠4 Given a list of student marks, write a program using reduce() to find the total marks and then compute the average.
marks=[54,98,47,87,35,75]
l1=reduce(lambda x,y:x+y,marks)
l2=l1/len(marks)
print(l2)

print(reduce(lambda x,y:x+y,marks)/len(marks))

```
