# Triple Layer Decorator

```python
import functools
def dec1(args1):
    def dec(func):
        @functools.wraps(func)
        def wrap1(*args,**kwargs):
            for i in range(args1):
                result=func(*args,**kwargs)
            return result
        return wrap1
    return dec
@dec1(3)
def say_hello():
    print("Hello! Abhi")
say_hello()
```

```python
# Decorators Questions
# 1. Create functions add(a, b), subtract(a, b) and multiply(a, b).
# Create a function calculate(operation, a, b) that accepts a function reference and performs the selected operation.
# Use lambda functions to perform:
# * Square of a number
# * Cube of a number
# * Double of a number
# Add a decorator log_operation that prints "Operation started" before execution and "Operation completed" after execution.
def log_operation(func):
    def wrap(*args,**kwargs):
        print("Operation Started")
        result=func(*args,**kwargs)
        print("Operation Completed")
        return result
    return wrap

def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mul(a,b):
    return a*b
@log_operation
def calculate(operation,a,b):
    return operation(a,b)
print(calculate(add,12,7))
print(calculate(sub,12,6))
print(calculate(mul,5,7))

square=lambda x:x*x
cube=lambda x:x**3
double=lambda x:x*2
print(square(5))
print(cube(5))
print(double(5))

```

```python
# 2 . Create a function process_marks(marks, operation) where operation is a function reference.
# Use lambda functions to:
# * Add 5 grace marks
# * Double each mark
# * Find whether a mark is greater than 40
# Create a decorator that prints "Processing started" and "Processing completed".
def dec(func):
    def wrap(*args,**kwargs):
        print("Processing started")
        result=func(*args,**kwargs)
        print("Processing Completed")
        return result
    return wrap
add=lambda x:x+5
double=lambda x:x*2
great=lambda x:x>40
@dec
def process_marks(marks,operation):
    return list(map(operation,marks))
marks=[45,78,35,98]
print(process_marks(marks,add))
print(process_marks(marks,double))
print(process_marks(marks,great))

```

```python
# 3. Create a function process_order(price, discount_function).
# Pass different lambda functions to calculate:
# * 10% discount
# * 20% discount
# * ₹100 flat discount
# Create two decorators:
# * order_logger → logs the order processing
# * payment_check → prints "Payment verification completed"
# Apply both decorators to the function.
def order_logger(func):
    def wrap1(*args,**kwargs):
        print("Order processing")
        r=func(*args,**kwargs)
        return r
    return wrap1
def Payment_check(func):
    def wrap2(*args,**kwargs):
        r=func(*args,**kwargs)
        print("Payment Verification Completed")
        return r
    return wrap2

dis_10=lambda x:x-(x*0.10)
dis_20=lambda x:x-(x*0.20)
dis_100=lambda x:x-100
@Payment_check
@order_logger
def process_order(price,discount_function):
    return discount_function(price)
print(process_order(1000,dis_10))
print(process_order(1000,dis_20))
print(process_order(1000,dis_100))
```

```python
# 4 . Create a function send_notification(message, formatter).
# Use lambda functions as formatter to:
# * Convert the message to uppercase
# * Convert the message to lowercase
# * Add "!!!" to the message
# Create a decorator that prints "Notification started" before execution and "Notification sent" after execution.
def dec(func):
    def wrap(*args,**kwargs):
        print("Notification Started")
        r=func(*args,**kwargs)
        print("Notifiacation Sent")
        return r
    return wrap

upper=lambda x:x.upper()
lower=lambda x:x.lower()
add=lambda x:x+"!!!"
@dec
def send_notification(message,formatter):
    return formatter(message)
print(send_notification("Hello, Abhi",upper))
print(send_notification("Hello, Abhi",lower))
print(send_notification("Hello, Abhi",add))
```
