# Generators

```python
# we use yield keyword in generator
# yield keyword as same as the return
# A generator is a special kind of iterator created with the regular function that uses yield keyword instead of return 
# Python handles all the iter and next machinery for you automatically
# Generators is also called as lazy iterators
```

```python
def m1():
    yield 10
    yield 20
    yield 30
    yield 40
x=m1()
for i in x:
    print(i)
x1=m1()
print()
print(next(x1))
print(next(x1))
print(next(x1))
print(next(x1))
# print(next(x1))     StopIteration: 
```

```python
def g2():
    n=1
    while True:
        yield n
        n+=1
y=g2()
print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
```

```python
def m1(start,stop):
    current=start
    while current<=stop:
        yield current
        current+=1
x=m1(1,5)
for i in x:
    print(i)
y=m1(5,10)
print()
print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
```

```python
def m1():
    n=1
    while True:
        yield n
        n+=1
x=m1()

for i in range(5):
    print(next(x))
```

```python
def m2():
    a,b=0,1
    while True:
        yield a
        a,b=b,a+b

x1=m2()
for i in range(5):
    print(next(x1))
print()
y=m2()
print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
```

```python
def m3():
    n=2
    while True:
        yield n
        n+=2

x=m3()
for i in range(5):
    print(next(x))
print()
y=m3()
print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
print(next(y))
```

```python
# list comprehension
l=[x for x in range(1,10)]
print(l)
print(type(l))

# Set
s={x for x in range(1,10)}
print(s)
print(type(s))

# dict
d={x:x*x for x in range(1,5)}
print(d)
print(type(d))

# tuple
t=tuple(x for x in range(1,10))
print(t)
print(type(t))

t1=(x for x in range(1,10))
for i in t1:
    print(i,end=" ")
print()

g=(x if x%2==0 else x+1 for x in range(1,10))
for i in g:
    print(i,end=" ")
print()

h=(x for i in range(2) for x in range(1,5))
for i in h:
    print(i,end=" ")
print()

j=(x for x in range(1,11) if x%2==0)
for i in j:
    print(i,end=" ")
print()

print(max(x for x in range(1,11) if x%2==0))
print(min(x for x in range(1,11) if x%2==0))
print(sum(x for x in range(1,11) if x%2==0))
```

# Generators Questions

```python
# 1.	Write a generator that yields numbers from 1 to N.
def numbers():
    n=1
    while True:
        yield n
        n+=1
x=numbers()
# print(next(x))

for _ in range(5):
    print(next(x))
```

```python
# 2.	Write a generator that yields even numbers from 1 to N
def even():
    n=2
    while True:
        yield n
        n+=2
x=even()
for _ in range(5):
    print(next(x))
```

```python
# 3.	Write a generator that yields each character of a string.
# def char(text):
#     for i in text:
#         yield char
# x=char("Abhi")

# print(next(x))

def characters(text):
    for char in text:
        yield char


text = input("Enter a string: ")

for char in characters(text):
    print(char)

```

```python
# 4.	Write a generator that yields characters of a string in reverse order.
def reverse(text):
    for i in text[::-1]:
        yield i
for i in reverse('abhilash'):
    print(i)
```

```python
# 5.	Write a generator that yields only vowels from a string.
def vowels(text):
    for i in text:
        if i in "aeiouAEIOU":
            yield i
for i in vowels("Akash"):
    print(i)
```

```python
# 6.	Write a generator that yields only digits present in a string.
def digit(text):
    for i in text:
        if i.isdigit():
            yield i

for i in digit("abc123xyz987"):
    print(i)
```

```python
# 7.	Write a generator that yields the square of each element in a list.
def square(numbers):
    for i in numbers:
        yield i*i

for i in square([1,4,7,9]):
    print(i)
```

```python
# 8.	Write a generator that yields digits from an integer one by one.
def dig(numbers):
    for i in str(numbers):
        yield i

for i in dig("12345"):
    print(i)

```

```python
# 9.	Create a generator that yields cumulative sum of numbers in a list. Example: [1,2,3] → 1, 3, 6
def cumulative_sum(numbers):
    total=0
    for i in numbers:
        total=total+i
        yield total
for i in cumulative_sum([1,2,3]):
    print(i)
```
