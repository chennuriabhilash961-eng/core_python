# Iterators and Generators

# Iterators

```python
# Iterators
# 1.iterable-iteration one after another
# 2.iterator - object
# it access the current value and the cursor moves to the next value
# calling the next it will access the next value and it moves the cursor to next value
# Iterators are single usage and it will go unidirectional
```

```python
# iterable
#-something you can loop over it 
#-Has an iter method that returns an iterator
#-Ex:list,tuple,string,dictionary,range,set

#iterator
#- something that produces one value at a time
#- It has both iter and next
```

```python
# iter() and next()
# An object that implements both iter() and next()
# methods correctly is an iterator

# iter()
#  def __iter__(self):  [method signature]
# An iter() must return an iterator object
# It is called by iter(obj)/obj.__iter__() and for loop

# next()
#  def __next__(self):  [method signature]
# It returns the next value and raises stopiteration when values are exhausted
# It can be called using next(obj)/obj.__next__() and for loop.
```

```python
# Custom iterators
class A:
    def __init__(self):
        self.val=1
    def __iter__(self):
        return self
    def __next__(self):
        if(self.val<=5):
            num=self.val
            self.val+=1
            return num
        else:
            raise StopIteration

obj=A()
# for i in obj:
#     print(i)

print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
print(obj.__next__())
# print(obj.__next__())   #StopIteration:
```

```python
class Numbers:
    def __init__(self,start,end):
        self.val=start
        self.end=end
    def __iter__(self):
        return self
    def __next__(self):
        if(self.val<=self.end):
            current=self.val
            self.val+=1
            return current
        else:
            raise StopIteration

obj=Numbers(5,10)
for i in obj:
    print(i)

obj1=Numbers(20,25)
for i in obj1:
    print(i)
        
```

```python
class Fibonacci:
    def __init__(self,n):
        self.count=1
        self.n=n
        self.a,self.b=0,1
    def __iter__(self):
        return self
    def __next__(self):
        if(self.count<=self.n):
            val=self.a
            self.a,self.b=self.b,self.a+self.b
            self.count+=1
            return val
        else:
            raise StopIteration

Fib1=Fibonacci(5)
for i in Fib1:
    print(i,end=" ")
print()
Fib2=Fibonacci(10)
for i in Fib2:
    print(i,end=" ")
 
```

```python
# Create an iterator class Countdown(n) that counts down from n to 0 (inclusive). After
# reaching 0, raise Stoplteration. Show that calling iter() on your iterator returns itself,
# and manually trace three next() calls.
class Countdown:
    def __init__(self,n):
        self.n=n
        self.count=0
    def __iter__(self):
        return self
    def __next__(self):
        if(self.n>0):
            val=self.n
            self.n-=1
            return val
        else:
            raise StopIteration

obj=Countdown(5)
for i in obj:
    print(i)
```

# Generators

```python
def m1(start,stop):
    current=start
    while current<=stop:
        yield current
        current+=1
x=m1(1,5)
# for i in x:
#     print(i)

print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
# print(next(x))     #StopIteration: 
```

```python
def g2():
    n=1
    while True:
        yield n
        n+=1
x=g2()
print(next(x))
```

```python
# 1.	Create an custom iterator that prints numbers from 1 to N, where N is given by the user.
class Numbers:
    def __init__(self,n):
        self.n=n
        self.count=1
    def __iter__(self):
        return self
    def __next__(self):
        if self.count<=self.n:
            num=self.count
            self.count+=1
            return num
        else:
            raise StopIteration

obj=Numbers(5)
for i in obj:
    print(i)
```

```python
# 2.	Create an custom iterator that prints numbers from N to 1.
class numbers:
    def __init__(self,n):
        self.n=n
        self.count=0
    def __iter__(self):
        return self
    def __next__(self):
        if self.n>0:
            val=self.n
            self.n-=1
            return val
        else:
            raise StopIteration

obj=numbers(10)
for i in obj:
    print(i)
            
```

```python
# 3.Create an custom iterator that prints the first N even numbers.
class Even:
    def __init__(self,n):
        self.n=n
        self.val=2
        self.count=0
    def __iter__(self):
        return self
    def __next__(self):
        if(self.count<self.n):
            num=self.val
            self.val+=2
            self.count+=1
            return num
        else:
            raise StopIteration

obj=Even(5)
for i in obj:
    print(i)
        
```

```python
# 5.	Create an custom iterator that returns only even numbers from a given list.
class even:
    def __init__(self,numbers):
        self.numbers=numbers
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        while self.index<len(self.numbers):
            num=self.numbers[self.index]
            self.index+=1
            if(num%2==0):
                return num
        raise StopIteration

obj=even([2,3,4,5,6,7,])
for i in obj:
    print(i)
        
```

```python
# 6.	Create an custom iterator that returns only odd numbers from a given list.
class odd1:
    def __init__(self,number):
        self.number=number
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        while self.index<len(self.number):
            num=self.number[self.index]
            self.index+=1
            if(num%2==1):
                return num
        raise StopIteration

obj=odd1([2,3,4,5,6,7])
for i in obj:
    print(i)
```

```python
# 7.	Create an custom iterator that returns only positive numbers from a list.
class Positive:
    def __init__(self,number):
        self.number=number
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        while self.index<len(self.number):
            num=self.number[self.index]
            self.index+=1
            if(num>0):
                return num
        raise StopIteration

obj=Positive([-1,2,-4,6,-2,9,-4])
for i in obj:
    print(i)
```

```python
# 8.	Create an custom iterator that prints each character of a string one by one.
class Character:
    def __init__(self,text):
        self.text=text
        self.index=0
    def __iter__(self):
        return self
    def __next__(self):
        while self.index<len(self.text):
            num=self.text[self.index]
            self.index+=1
            return num
        raise StopIteration

obj=Character("Shravan")
for i in obj:
    print(i)
        
```

```python
# 9.	Create an custom iterator that prints the characters of a string in reverse order.
class Reverse:
    def __init__(self,text):
        self.text=text
        self.index=len(text)-1
    def __iter__(self):
        return self
    def __next__(self):
        if self.index>=0:
            num=self.text[self.index]
            self.index-=1
            return num
        else:
            raise StopIteration
        
obj=Reverse("abhilash")
for i in obj:
    print(i)
```
