# Recursion

```python
# Recursion is a programming technique where a function calls itself either directly or indirectly to solve the problem.It is commonly used for 
# 1 breaking problems into smaller sub problems
# 2 Mathematical calculations like factorial and fibonacci
def greet():
    print("Hi")
    greet()
greet()
```

```python
def greet(n):
    if(n>5):
        return None
    print("Hi")
    greet(n+1)
greet(1)
```

```python
def m1(n):
    if(n>5):
        return None
    print(n)
    m1(n+1)
m1(1)
```
